```python
from __future__ import annotations

from pathlib import Path
from typing import Callable

import matplotlib.pyplot as plt
from qiskit import QuantumCircuit, transpile
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_histogram
from qiskit.result.counts import Counts
from qiskit_aer import AerSimulator

SHOTS = 1024

SAVE_FOLDER = Path.cwd() / "qcl_output"
SAVE_FOLDER.mkdir(exist_ok=True)

def create_superposition(qubits: int = 1) -> QuantumCircuit:
    """
    Creates a circuit where every qubit is placed
    into an equal superposition using Hadamard gates.
    """
    circuit = QuantumCircuit(qubits, name="Superposition")

    for i in range(qubits):
        circuit.h(i)

    return circuit

def create_bell_state() -> QuantumCircuit:
    """
    Creates a 2-qubit Bell State.

    Circuit:
            H(q0)
            CX(q0,q1)
    """

    circuit = QuantumCircuit(2, name="Bell State")

    circuit.h(0)
    circuit.cx(0, 1)

    return circuit

def create_ghz_state(qubits: int = 3) -> QuantumCircuit:
    """
    Creates an N-qubit GHZ State.

    Example (3 qubits):

        |000>
              |
                  H
                    |
                          CNOT
                                |
                                      CNOT
    """

    if qubits < 2:
        raise ValueError("GHZ State requires at least two qubits.")

    circuit = QuantumCircuit(qubits, name="GHZ State")

    circuit.h(0)

    for target in range(1, qubits):
        circuit.cx(0, target)

    return circuit

def execute_circuit(circuit: QuantumCircuit,
                    shots: int = SHOTS) -> Counts:
    """
    Executes the given circuit on the Aer simulator
    after adding measurement operations.
    """

    simulator = AerSimulator()

    measured = circuit.copy()
    measured.measure_all()

    compiled = transpile(measured, simulator)

    result = simulator.run(compiled, shots=shots).result()

    return result.get_counts()

def print_statevector(circuit: QuantumCircuit) -> Statevector:
    """
    Prints amplitudes and probabilities of every basis state.
    """

    state = Statevector(circuit)

    print("\nStatevector before measurement\n")

    for state_label, amplitude in state.to_dict().items():

        probability = abs(amplitude) ** 2

        print(
            f"|{state_label}>"
            f"\tAmplitude = {amplitude.real:+.4f}"
            f"{amplitude.imag:+.4f}j"
            f"\tProbability = {probability*100:.2f}%"
        )

    return state

def display_histogram(counts: Counts,
                      title: str) -> None:
    """
    Displays and saves the measurement histogram.
    """

    print("\nMeasurement Results")
    print("-" * 35)

    total = sum(counts.values())

    for outcome in sorted(counts):

        percentage = counts[outcome] * 100 / total

        print(
            f"{outcome:>5} : "
            f"{counts[outcome]:4d} shots "
            f"({percentage:.2f}%)"
        )

    figure = plot_histogram(counts, title=title)

    try:
        filename = SAVE_FOLDER / f"{title.lower().replace(' ','_')}.png"
        figure.savefig(filename, bbox_inches="tight")
        print(f"\nHistogram saved to:\n{filename}")

    except OSError as error:
        print("Unable to save histogram:", error)

    try:
        plt.show()
    except Exception:
        pass

    plt.close(figure)

EXPLANATIONS = {

    "Superposition":

        "Hadamard gates place each qubit into an equal mixture "
        "of |0> and |1>. Each measurement randomly collapses "
        "to either state with approximately equal probability.",

    "Bell State":

        "Applying a Hadamard followed by a CNOT creates an "
        "entangled pair. Only |00> and |11> should appear "
        "during measurement.",

    "GHZ State":

        "A GHZ state extends Bell-state entanglement to multiple "
        "qubits. Measurements should produce either all zeros "
        "or all ones."
}

def run_demo(name: str,
             circuit: QuantumCircuit) -> None:
    """
    Executes every step required to demonstrate
    a quantum circuit.
    """

    print("\n" + "=" * 65)
    print(f"{name} ({circuit.num_qubits} qubits)")
    print("=" * 65)

    print("\nCircuit Diagram\n")
    print(circuit.draw(output="text"))

    print_statevector(circuit)

    counts = execute_circuit(circuit)

    display_histogram(counts, title=name)

    print("\nTheory")
    print("-" * 30)
    print(EXPLANATIONS[name])

MENU = """
=============================
 Quantum Circuit Library
 =============================

 1. Superposition

 2. Bell State

 3. GHZ State

 4. Exit

 """

MENU_OPTIONS: dict[str, tuple[str, Callable[[], QuantumCircuit]]] = {

    "1": ("Superposition", create_superposition),

    "2": ("Bell State", create_bell_state),

    "3": ("GHZ State", create_ghz_state)

}

def main():

    print("\nWelcome to the Quantum Circuit Library!\n")

    while True:

        print(MENU)

        choice = input("Enter your choice (1-4): ").strip()

        if choice == "4":

            print("\nThank you for using the program!")

            break

        if choice not in MENU_OPTIONS:

            print("\nInvalid option! Please try again.\n")

            continue

        title, function = MENU_OPTIONS[choice]

        run_demo(title, function())

        input("\nPress Enter to continue...")

if __name__ == "__main__":
    main()
```
